```python


import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Base de datos de hoteles
# características: costo, ubicación, estrellas
# costo: 0=bajo, 1=medio, 2=alto
# ubicación: 0=lejos, 1=cerca
# estrellas: 1 al 5

hoteles = {
    0: {"nombre": "Hotel Barato",   "caract": [0, 0, 2]},
    1: {"nombre": "Hotel Centro",   "caract": [1, 1, 3]},
    2: {"nombre": "Hotel Lujo",     "caract": [2, 1, 5]},
    3: {"nombre": "Hotel Playa",    "caract": [1, 0, 4]},
    4: {"nombre": "Hotel Express",  "caract": [0, 1, 1]},
    5: {"nombre": "Hotel Familiar", "caract": [1, 1, 4]},
    6: {"nombre": "Hotel Rural",    "caract": [0, 0, 3]}
}
# 2. Datos de entrenamiento
X_train = np.array([
    [0, 0, 2],  # barato, lejos, 2 estrellas - Hotel Barato
    [1, 1, 3],  # medio, cerca, 3 estrellas - Hotel Centro
    [2, 1, 5],  # caro, cerca, 5 estrellas - Hotel Lujo
    [1, 0, 4],  # medio, lejos, 4 estrellas - Hotel Playa
    [0, 1, 1],  # barato, cerca, 1 estrella - Hotel Express
    [1, 1, 4],  # medio, cerca, 4 estrellas - Hotel Familiar (nuevo)
    [0, 0, 3],  # barato, lejos, 3 estrellas - Hotel Rural (nuevo)
    [1, 0, 3],  # medio, lejos, 3 estrellas - Hotel Rural (nuevo escenario)
    [2, 1, 4]   # caro, cerca, 4 estrellas - Hotel Lujo (por cercanía)
], dtype=np.float32)

y_train = np.array([0, 1, 2, 3, 4, 5, 6, 6, 2])

#Construcción del modelo de red neuronal
model = keras.Sequential([
    layers.Dense(8, activation='relu', input_shape=(3,)),
    layers.Dense(5, activation='relu'),
    layers.Dense(7, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

#Entrenamiento (verificacion en caso de error)
print("Entrenando modelo...")
model.fit(X_train, y_train, epochs=500, verbose=0)
print("Entrenamiento completado.\n")

#Función para recomendar hotel
def recomendar_hotel(preferencias):
    entrada = np.array([preferencias], dtype=np.float32)
    prediccion = model.predict(entrada, verbose=0)
    indice_hotel = np.argmax(prediccion[0])
    return indice_hotel, hoteles[indice_hotel]["nombre"]

#Interacción con el usuario
print("SISTEMA DE RECOMENDACIÓN DE HOTELES")
print("Responde las siguientes preguntas:")

costo = int(input("Costo (0=bajo, 1=medio, 2=alto): "))
ubicacion = int(input("Ubicación (0=lejos del centro, 1=cerca): "))
estrellas = int(input("Estrellas mínimas (1 a 5): "))

if estrellas > 5:
    estrellas = 5

pref_usuario = [costo, ubicacion, estrellas]

idx, nombre = recomendar_hotel(pref_usuario)
print(f"\nHotel recomendado: {nombre} (índice {idx})")
```
